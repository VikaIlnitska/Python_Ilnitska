## Завдання 1.

$$z_1 = \cos\alpha + \cos 2\alpha + \cos 6\alpha + \cos 7\alpha$$
$$z_2 = 4\cos\frac{\alpha}{2} \cdot \cos\frac{5}{2}\alpha \cdot \cos 4\alpha$$

```python
import numpy as np

alpha = np.pi / 6

z1 = np.cos(alpha) + np.cos(2*alpha) + np.cos(6*alpha) + np.cos(7*alpha)
z2 = 4 * np.cos(alpha/2) * np.cos(5/2*alpha) * np.cos(4*alpha)

print(f"alpha = π/6")
print(f"z1 = {z1:.2f}")
print(f"z2 = {z2:.2f}")
```

## Завдання 2.

$$\ln(1-x) = -\sum_{n=1}^{\infty}\frac{x^n}{n} = -\left(x +\frac{x^2}{2}+\frac{x^4}{4}+ ...\right)\quad-1\leq x < 1$$

```python
import numpy as np

x = 0.5
precision = 10

result_math = np.log(1 - x)

s_sum = 0
n = 1
while True:
    term = (x**n) / n
    s_sum += term
    if round(-s_sum, precision) == round(result_math, precision):
        break
    n += 1

result_s = -s_sum

print(f"x = {x}")
print(f"Точність: {precision} (знаків після коми)")
print(f"Значення пораховане першим способом: {result_math:.{precision}f}")
print(f"Значення пораховане другим способом: {result_s:.{precision}f}")
print(f"Кількість підсумованих членів ряду: {n}")
```

## Завдання 3.

Дано цілочислову прямокутну матрицю розмірності 13x13. Матрицю задати в коді програми. Генератор випадкових чисел не використовувати. 
Визначити:
1) суму елементів у тих рядках, які містять хоча б один від'ємний елемент;
2) номери рядків і стовпців усіх сідлових точок матриці.
3) *Примітка* Матриця А має сідлову точку A_ij, якщо A_ij є мінімальним елементом в i-му рядку та максимальним у j-му стовпці.

```python
import numpy as np

matrix = np.array([
    [ 5,  6,  7,  8,  9,  6,  7,  8,  9,  6,  7,  8,  9],
    [ 4, -1,  4,  1,  5, -9,  2,  6,  5,  3, -5,  8,  9],
    [ 3,  9, -3,  2,  3,  8, -4,  6,  2,  7,  1,  9,  5],
    [ 2,  2,  8,  4,  6,  7,  1, -3,  9,  5,  2,  8,  4],
    [ 1,  6,  2, -8,  5,  3,  9,  7,  4,  1,  6,  2,  8],
    [ 4,  5, -1,  9,  2,  6,  3, -7,  8,  4,  5,  1,  9],
    [ 3,  7,  1,  5, -3,  4,  6,  8,  1,  2,  7,  5,  3],
    [ 2,  3,  9, -2,  7,  1,  5,  4,  6,  8,  3,  9,  2],
    [ 1,  4,  7,  3, -1,  9,  2,  5,  8,  6,  4,  7,  1],
    [ 4,  8, -4,  6,  1,  5,  7,  2,  3,  9,  8,  4,  6],
    [ 3,  1,  6,  7,  4, -2,  8,  3,  1,  5,  1,  6,  7],
    [ 2,  6,  5,  1,  8,  7, -4,  9,  2,  3,  6,  5,  1],
    [ 1,  2,  3,  4,  9, -6,  1,  8,  5,  7,  2,  3,  4]
])

row_sums = 0
for row in matrix:
    if np.any(row < 0):
        row_sums += np.sum(row)

print("1) Сума елементів рядків з від'ємними елементами:", row_sums)

saddle_points = []
for i in range(matrix.shape[0]):
    row_min = np.min(matrix[i, :])
    for j in range(matrix.shape[1]):
        if matrix[i, j] == row_min:
            col_max = np.max(matrix[:, j])
            if matrix[i, j] == col_max:
                saddle_points.append((i, j))

if saddle_points:
    print("2) Сідлові точки матриці:", ", ".join([f"Рядок {p[0]+1} Стовпець {p[1]+1} Значення {matrix[p[0], p[1]]}" for p in saddle_points]))
else:
    print("2) Сідлових точок не знайдено")
```
